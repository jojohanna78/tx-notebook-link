# Comparaison des libraires de visualisation

## Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

def plot_sin(freq=1.0):
    x = np.linspace(0, 2*np.pi, 200)
    plt.figure(figsize=(6, 4))
    plt.plot(x, np.sin(freq * x))
    plt.title(f"Sinus (matplotlib), fréquence = {freq}")
    plt.show()

interact(plot_sin, freq=(0.5, 5.0, 0.5))
```

## Plotly

```python
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display
import numpy as np

x = np.linspace(0, 2*np.pi, 200)

fig = go.FigureWidget(data=go.Scatter(x=x, y=np.sin(x), mode='lines'))
fig.update_layout(title="Sinus interactif (plotly)", xaxis_title="x", yaxis_title="sin(x)")

slider = widgets.FloatSlider(min=0.5, max=5.0, step=0.5, value=1.0, description='Fréquence:')

def on_change(change):
    fig.data[0].y = np.sin(change['new'] * x)

slider.observe(on_change, names='value')
display(slider, fig)
```

## Bokeh

```python
from bokeh.plotting import figure, show
from bokeh.models import CustomJS, Slider, ColumnDataSource
from bokeh.layouts import column
from bokeh.io import output_notebook
import numpy as np

output_notebook()

x = np.linspace(0, 2*np.pi, 200)
source = ColumnDataSource(data=dict(x=x, y=np.sin(x)))

p = figure(title="Sinus interactif (bokeh, CustomJS)", x_axis_label='x', y_axis_label='sin(x)')
p.line('x', 'y', source=source, line_width=2)

slider = Slider(start=0.5, end=5.0, value=1.0, step=0.5, title="Fréquence")

callback = CustomJS(args=dict(source=source, freq=slider), code="""
    const data = source.data;
    const f = freq.value;
    const x = data['x'];
    const y = data['y'];
    for (let i = 0; i < x.length; i++) {
        y[i] = Math.sin(f * x[i]);
    }
    source.change.emit();
""")

slider.js_on_change('value', callback)

show(column(slider, p))
```
