# Widget examples

## Le slider 

Il permet de faire glisser un curseur entre plusieurs valeurs numériques, qui sont ensuite stockées dans une variable. On utilise ici la méthode .observe() pour détecter chaque changement de valeur et déclencher une action en conséquence. Ce mécanisme d'écoute est la base de la plupart des interactions avec les widgets : on associe une fonction de rappel (callback) à un événement de changement.

```python
import ipywidgets as widgets
from IPython.display import display
slider = widgets.IntSlider(min=0, max=100, value=50, description='Valeur:')
output = widgets.Output()

def on_change(change):
    with output:
        output.clear_output()
        print(f"Nouvelle valeur : {change['new']}")

slider.observe(on_change, names='value')
display(slider, output)
```

## Dropdown

Il permet de choisir une valeur parmi une liste d'options prédéfinies via un menu déroulant. C'est un widget qui est adapté quand on veut restreindre l'utilisateur à un ensemble fini de choix, par exemple pour sélectionner un jeu de données, une catégorie ou un paramètre parmi plusieurs valeurs possibles.

```python
dropdown = widgets.Dropdown(
    options=['Banane', 'Carotte', 'Fraise'],
    value='Banane',
    description='Choix:'
)
display(dropdown)
```

## Text

Permet à l'utilisateur de saisir librement du texte dans un champ de saisie. Ce widget est utile pour récupérer une entrée personnalisée (un nom, une recherche, un paramètre textuel).

```python
texte = widgets.Text(
    value='',
    placeholder='Tape quelque chose...',
    description='Nom:'
)
display(texte)
```

## Checkbox

Permet d'activer ou de désactiver une option grâce à une case à cocher. Il est utilisé pour afficher ou masquer un élément, activer une option facultative, ou basculer entre deux modes d'affichage.

```python
case = widgets.Checkbox(
    value=False,
    description='Cocher pour activer'
)
display(case)
```

## Button

Contrairement aux widgets précédents qui réagissent à un changement de valeur, le bouton réagit à un clic. On associe une fonction à l'événement `on_click`, qui s'exécute à chaque fois que l'utilisateur clique dessus. C'est le widget est utile quand on veut lancer une action ponctuelle (lancer un calcul, valider un formulaire, réinitialiser un affichage).

```python
bouton = widgets.Button(description='Cliquez-moi')
sortie = widgets.Output()

def on_click(b):
    with sortie:
        print("Bouton cliqué !")

bouton.on_click(on_click)
display(bouton, sortie)
```

## Output

C'est une zone d'affichage dynamique, généralement utilisée en complément d'un autre widget. Elle capture tout ce qui est imprimé (print, graphiques, résultats…) à l'intérieur du bloc `with output:`, et permet de rafraîchir son contenu avec `.clear_output()` sans recréer toute la cellule. C'est le widget qu'on retrouve dans presque tous les exemples précédents, puisqu'il sert de support d'affichage aux résultats produits par les interactions.

```python
sortie = widgets.Output()

with sortie:
    print("Ceci s'affiche dans la zone Output")

display(sortie)
```

## HTML

Permet d'afficher du contenu HTML/CSS personnalisé directement dans le notebook. Ce widget est pratique pour styliser un titre, une mise en forme ou un message, ou pour intégrer du contenu plus riche (mise en couleur, mise en page).

```python
html_widget = widgets.HTML(
    value="<h3 style='color:steelblue'>Titre </h3><p>Texte en <b>gras</b>.</p>"
)
display(html_widget)
```

## Exemple de combinaison de widgets pour modéliser une courbe

Les widgets présentés précédemment ne sont pas limités à un usage isolé : leur intérêt principal est dans leur capacité à piloter dynamiquement un autre élément du notebook, typiquement un graphique. En associant plusieurs widgets de saisie à une même fonction de rappel via `.observe()`, on peut modifier en temps réel les données affichées, sans avoir à réexécuter manuellement la cellule.

C'est ce principe qui permet, par exemple, de combiner un Dropdown et un slider pour faire varier à la fois le type de fonction affichée (sinus, cosinus, tangente) et sa fréquence : chaque widget ne fait qu'exposer une valeur, et c'est une fonction unique qui se charge de relire l'état des deux contrôles et de mettre à jour le graphique en conséquence.

```python
import numpy as np
import bqplot.pyplot as bplt
import ipywidgets as widgets
from IPython.display import display

x = np.linspace(0, 2*np.pi, 200)

fonctions = {'Sinus': np.sin, 'Cosinus': np.cos, 'Tangente': np.tan}

fig = bplt.figure(title="Fonction + fréquence")
courbe = bplt.plot(x, np.sin(x))

dropdown = widgets.Dropdown(options=list(fonctions.keys()), value='Sinus', description='Fonction:')
slider = widgets.FloatSlider(min=0.5, max=5.0, step=0.5, value=1.0, description='Fréquence:')

def update(change=None):
    f = fonctions[dropdown.value]
    courbe.y = f(slider.value * x)

dropdown.observe(update, names='value')
slider.observe(update, names='value')

update()  

display(widgets.HBox([dropdown, slider]), fig)
```
