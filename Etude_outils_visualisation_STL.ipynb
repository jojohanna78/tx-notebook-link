# Etude d’outils de visualisation de fichiers STL

## Première piste : trimesh + matplotlib

```python
import trimesh
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Charger le fichier STL (remplacez par le chemin de votre fichier)
mesh = trimesh.load('storage.STL')

# Créer une figure 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Extraire les données 
vertices = mesh.vertices
faces = mesh.faces

# Tracer la mesh 
ax.plot_trisurf(vertices[:, 0], vertices[:, 1], vertices[:, 2], 
                triangles=faces, 
                cmap='viridis', 
                linewidth=0.2, 
                antialiased=True,
                shade=True)

# Désactiver les axes
plt.axis('off')
plt.tight_layout()

# Sauvegarder la figure et l'afficher
plt.savefig('temp_plot.png', dpi=75, bbox_inches='tight')
plt.close()

# Afficher l'image sauvegardée
from IPython.display import Image
Image(filename='temp_plot.png')
```

## Deuxième piste : matplotlib + ipywidgets 

```python
import trimesh
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
from ipywidgets import interact, FloatSlider

# Entrez le chemin de votre fichier STL
mesh = trimesh.load('Cylinder.stl')

# Fonction pour créer la visualisation interactive
def visualisation_interactive(elevation=30, azimuth=45, zoom=1.0):
    """
    Visualisation 3D interactive du fichier STL
    Paramètres:
    - elevation: angle d'élévation (de 0 à 90 degrés)
    - azimuth: angle d'azimut (de 0 à 360 degrés)
    - zoom: niveau de zoom (de 0.5 à 2.0)
    """
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Extraire les données 
    vertices = mesh.vertices
    faces = mesh.faces
    
    # Tracer la mesh
    ax.plot_trisurf(vertices[:, 0], vertices[:, 1], vertices[:, 2], 
                    triangles=faces, 
                    cmap='viridis', 
                    linewidth=0.2, 
                    antialiased=True,
                    shade=True)
    
    # Désactiver les axes 
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.set_zticklabels([])
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_zticks([])
    plt.axis('off')
    
    # Ajuster la vue selon les paramètres
    ax.view_init(elev=elevation, azim=azimuth)
    
    # Calculer les limites en fonction du zoom
    max_range = np.array([vertices[:,0].max()-vertices[:,0].min(), 
                          vertices[:,1].max()-vertices[:,1].min(), 
                          vertices[:,2].max()-vertices[:,2].min()]).max() / 2.0
    zoom_factor = 1.0 / zoom
    
    mid_x = (vertices[:,0].max()+vertices[:,0].min()) * 0.5
    mid_y = (vertices[:,1].max()+vertices[:,1].min()) * 0.5
    mid_z = (vertices[:,2].max()+vertices[:,2].min()) * 0.5
    
    ax.set_xlim(mid_x - max_range * zoom_factor, mid_x + max_range * zoom_factor)
    ax.set_ylim(mid_y - max_range * zoom_factor, mid_y + max_range * zoom_factor)
    ax.set_zlim(mid_z - max_range * zoom_factor, mid_z + max_range * zoom_factor)
    
    plt.tight_layout()
    plt.show()

# Créer l'interface interactive
interact(visualisation_interactive,
         elevation=FloatSlider(min=0, max=90, step=5, value=30, description='Élévation'),
         azimuth=FloatSlider(min=0, max=360, step=5, value=45, description='Azimut'),
         zoom=FloatSlider(min=0.5, max=2.0, step=0.1, value=1.0, description='Zoom'))
```
